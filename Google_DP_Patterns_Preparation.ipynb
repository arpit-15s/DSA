# 📘 Google Interview Prep: Dynamic Programming (DP) Patterns

Prepare these essential DP patterns for your Google Software Engineer interview.

---

## ✅ 1. Fibonacci Pattern (Basic Recursion + Memoization)
**Concept:** Top-down recursion with memoization or bottom-up tabulation.

**Examples:**
- Fibonacci series
- Climbing stairs
- House robber
- Decode ways

## ✅ 2. 0/1 Knapsack Pattern
**Concept:** Choosing to take or not take an item with a capacity constraint.

**Examples:**
- 0/1 Knapsack
- Subset sum
- Target sum
- Partition equal subset sum

## ✅ 3. Unbounded Knapsack Pattern
**Concept:** You can take an item **multiple times**.

**Examples:**
- Coin change (min number of coins)
- Rod cutting
- Integer break

## ✅ 4. Longest Common Subsequence (LCS) Pattern
**Concept:** Matching subsequences between two sequences.

**Examples:**
- LCS
- Longest Palindromic Subsequence
- Edit Distance
- Shortest Common Supersequence

## ✅ 5. Matrix Chain Multiplication (MCM) / Interval DP
**Concept:** Range-based optimization problems.

**Examples:**
- Matrix Chain Multiplication
- Burst Balloons
- Palindrome Partitioning
- Boolean Parenthesization

## ✅ 6. Palindrome DP
**Concept:** Checking or partitioning substrings/subarrays for palindromes.

**Examples:**
- Palindrome Partitioning
- Longest Palindromic Subsequence
- Longest Palindromic Substring

## ✅ 7. Digit DP (Advanced)
**Concept:** Counting or optimizing over digit constraints.

**Examples:**
- Count numbers with unique digits < N
- Count numbers divisible by K with constraints

## ✅ 8. Bitmask DP
**Concept:** DP over subsets using bit representation.

**Examples:**
- Traveling Salesman Problem (TSP)
- Partition to K equal subsets
- Minimum cost to connect cities

## ✅ 9. DP on Trees / Graphs
**Concept:** Dynamic Programming where the input is a tree or graph.

**Examples:**
- Tree Diameter
- Max path sum in binary tree
- House Robber III

## ✅ 10. Game Theory DP
**Concept:** Turn-based strategy or winning condition evaluation.

**Examples:**
- Stone Game
- Predict the Winner
- Can I Win?

## ✅ 11. Greedy + DP Hybrid
**Concept:** Combining greedy observations with dynamic programming.

**Examples:**
- Jump Game II
- Minimum number of deletions to make a string palindrome
