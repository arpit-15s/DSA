# Microsoft Interview Prep (LLD + HLD) - 7 Day Crash Plan

## 🔹 Day 1 – Core LLD Foundations
- Revise **OOP** (Inheritance, Polymorphism, Encapsulation, Abstraction).
- **SOLID principles** with examples.
- Revise **top 6 design patterns** (Singleton, Factory, Builder, Strategy, Observer, Decorator).
- Practice **Parking Lot design** (class diagram + extensibility).

## 🔹 Day 2 – More LLD Problems
- Design **Library Management System**.
- Design **Elevator System**.
- Learn **thread-safe Singleton & Producer-Consumer (multithreading basics)**.
- Deliverable: UML diagrams + extensibility discussion.

## 🔹 Day 3 – HLD Basics
- **Scalability building blocks**: Load balancer, Cache, DB replication/sharding, Message Queues.
- **SQL vs NoSQL**, CAP theorem, indexing.
- **API design best practices**.
- Deliverable: Architecture diagram for a small system (e.g., URL Shortener).

## 🔹 Day 4 – HLD Small Systems
- **Design URL Shortener (bit.ly)**.
- **Design Rate Limiter**.
- Focus: APIs, DB schema, architecture, trade-offs.

## 🔹 Day 5 – HLD Medium Systems
- **Design Chat System (WhatsApp-lite)**.
- **Design Notification Service (Email/SMS/Push)**.
- Focus: Asynchronous flows (Kafka/SQS), scaling storage, caching.

## 🔹 Day 6 – Mock + Tough Scenarios
- Pick **1 LLD**: *Splitwise* or *Food Delivery System*.
- Pick **1 HLD**: *News Feed* or *Ride Sharing*.
- Mock Interview Flow: Requirements → APIs → DB schema → Architecture → Trade-offs.

## 🔹 Day 7 – Revision & Polishing
- Revise all practiced diagrams.
- Prepare **2-minute crisp intros** for each design.
- Review **common mistakes**: jumping into design, ignoring NFRs.
- Practice **trade-off discussions** (SQL vs NoSQL, push vs pull, cache eviction).

---
## 📌 Microsoft Specific Tips
- Expect **LLD questions first** → focus on coding + design thinking.
- For HLD → clarity + trade-offs matter more than exotic scaling.
- Be prepared for **follow-ups**: e.g., *"how will you scale if users grow 10x?"* → always 2 options ready.
